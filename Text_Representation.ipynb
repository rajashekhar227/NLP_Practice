**Removing stop words**

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words=set(stopwords.words('english'))
```

```python
df_temp=df
```

```python
df.head()
```

```python
def remove_stopwords(text):
    words = text.split()
    rem_words = [word for word in words if word.lower() not in stop_words]
    return ' '.join(rem_words)
```

```python
df['summary']=df['summary'].apply(remove_stopwords)
```

# Tokenization

```python
import spacy
```

```python
tok=spacy.load('en_core_web_sm')
def tokenization(sentence):
  doc=tok(sentence)
  return [token.text for token in doc]
```

```python
df['summary_tokens']=df['summary'].apply(tokenization)
```
